```python
import subprocess, sys, shlex
from pathlib import Path

CONFIG = "configs/instant-mesh-base.yaml"
INPUT_DIR = "outputs_sam2/image2_overlay.jpg"  
OUTPUT_DIR = "outputs2"                       # or any path you like
SAVE_VIDEO = True                            # set False if you don't want the .mp4
EXPORT_TEXMAP = True                        # True -> saves textured mesh (.mtl + texture)

# Build the CLI command (single run handles all images in the folder)
cmd = [
    sys.executable, "run.py",
    CONFIG,
    INPUT_DIR,
    "--output_path", OUTPUT_DIR,
    "--diffusion_steps", "50",
    "--seed", "42",
    "--scale", "1.0",
    "--distance", "4.5",
    "--view", "4",
]
if SAVE_VIDEO:
    cmd.append("--save_video")
if EXPORT_TEXMAP:
    cmd.append("--export_texmap")

print("Running:", " ".join(shlex.quote(c) for c in cmd))
subprocess.run(cmd, check=True)
```
