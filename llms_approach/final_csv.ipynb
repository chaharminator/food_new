```python
from pathlib import Path
import pandas as pd

data_dir = Path.cwd()
cols_to_use = [0, 1, 3, 4, 5]
col_names = {0: "dish_id", 1: "calories", 3: "fat", 4: "carbs", 5: "protein"}

predictions_path = data_dir / "foodqwen_macro_estimates.csv"
if not predictions_path.exists():
    raise FileNotFoundError("Missing foodqwen_macro_estimates.csv in the working directory.")

cafe1 = (
    pd.read_csv(
        data_dir / "nutrition5k_dataset_metadata_dish_metadata_cafe1.csv",
        header=None,
        usecols=cols_to_use,
    ).rename(columns=col_names)
)
cafe2 = (
    pd.read_csv(
        data_dir / "nutrition5k_dataset_metadata_dish_metadata_cafe2.csv",
        header=None,
        usecols=cols_to_use,
    ).rename(columns=col_names)
)

groundtruth_all = pd.concat([cafe1, cafe2], ignore_index=True)

predictions = pd.read_csv(predictions_path)
prediction_ids = predictions["dish_id"].astype(str).unique()

groundtruth_filtered = groundtruth_all[groundtruth_all["dish_id"].astype(str).isin(prediction_ids)].reset_index(drop=True)

output_path = data_dir / "nutrition5k_dataset_groundtruth.csv"
groundtruth_filtered.to_csv(output_path, index=False)

output_path
```
